# **Pandas Library**

Pandas is the most popular open-source Python library used for data manipulation, analysis, and cleaning. Built directly on top of NumPy, it provides fast and flexible tools to work with structured, tabular data like spreadsheets or SQL tables.

```python
import pandas as pd
```

Pandas Series and Pandas DataFrames are the two primary data structures used in Python's Pandas library for data cleaning, manipulation, and analysis. The core difference is that a Series represents a single column (1D array), while a DataFrame represents an entire table (2D array) made up of multiple columns.

## **Quick Comparison: Series vs. DataFrame**

When working with Pandas, it's crucial to understand the difference between its two core data structures. 

| Feature | `Pandas Series` | `Pandas DataFrame` |
| :--- | :--- | :--- |
| **Dimensions** | 1D (Single column) | 2D (Rows & Columns) |
| **Data Types** | Homogeneous (Usually one type per Series) | Heterogeneous (Different types per column) |
| **Analogy** | A single column in Excel | An entire Excel spreadsheet |
| **Creation** | Lists, dictionaries, or 1D arrays | Dictionary of arrays, lists of dicts, or CSVs |

## **Pandas Series**

A **Series** is the fundamental building block of Pandas. It is a one-dimensional array capable of holding any data type. Notice how Pandas automatically generates a numeric index (0, 1, 2...) next to your data!

```python
temparatures = pd.Series([32, 45, 28, 30, 35])
print(temparatures)
print(type(temparatures))
```

```python
temparatures = pd.Series([32, 45, 28, 30, 35], index=["Monday","Tuesday","Wednesday","Thrusday","Friday"])
print(temparatures)
```

## **`.loc` and `.iloc`**

- The primary difference between .loc and .iloc in a Pandas DataFrame is that .loc uses label-based indexing to select data by names, while .iloc uses integer-based indexing to select data by numerical positions. Both attributes use the identical syntax format of df.method[row_indexer, column_indexer] to look up or slice your data.

- When slicing and selecting data in a Pandas DataFrame, understanding the difference between label-based and position-based indexing will save you from constant errors. 

| Feature | `.loc` | `.iloc` |
| :--- | :--- | :--- |
| **Indexing Basis** | Custom Row/Column labels (names). | Integer positions (0-indexed). |
| **Slicing Endpoint** | **Inclusive** (both start and stop labels appear). | **Exclusive** (stops right before the end index). |
| **Boolean Filtering** | Supported natively (e.g., `df['Age'] > 30`). | Not supported directly with boolean Series. |
| **Missing Keys** | Raises a `KeyError`. | Raises an `IndexError` if out-of-bounds. |

```python
temparatures = pd.Series([32, 45, 28, 30, 35], index=["Monday","Tuesday","Wednesday","Thrusday","Friday"])
print(temparatures.iloc[4])
print(temparatures.loc["Friday"])
```

## **Pandas DataFrame**

A Pandas DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). It is the primary data structure used in the Python Pandas library, functioning much like an Excel spreadsheet or a SQL table.

- Key Characteristics
    - **Two-Dimensional**: Data is organized cleanly into rows and columns.
    - **Heterogeneous**: Different columns can hold completely different data types (e.g., strings, integers, dates).
    - **Labeled Axes**: Rows have an index and columns have names for direct, semantic referencing.


```python
data = {
    "Name" : ["Alice", "Bob", "Charlie"],
    "Age" : [20, 25, 30],
    "Gender" : ["F", "M", "M"]
}

dt = pd.DataFrame(data, index=["Student_1", "Student_2", "Student_3"])
print(dt)
print(type(dt))
```

```python
data = {
    "Name" : ["Alice", "Bob", "Charlie"],
    "Age" : [20, 25, 30],
    "Gender" : ["F", "M", "M"]
}

dt = pd.DataFrame(data, index=["Student_1", "Student_2", "Student_3"])

# Accessing data using iloc and loc
print(f"iloc in 30 : {dt.iloc[2, 1]}")
print(f"loc in 30: {dt.loc["Student_3","Age"]}\n")

# Accessing data in Rows
print(f"Student_2 row : \n{dt.iloc[1]}")
print(f"type of Student_2 row : {type(dt.iloc[1])}\n")

# Accessing data in Colomns
print(f"Age Colomn : \n{dt["Age"]}")
print(f"type of Age Colomn : {type(dt["Age"])}\n")

print(f"Shape of dt : {dt.shape}")
print(f"Size of dt : {dt.size}\n")
print(f"len of dt(Number of Rows) :  {len(dt)}\n")  #Number of Rows
print(f"dTypes of dt : \n{dt.dtypes}\n")
```

## **1. The `.sum()` Method**
* **What it does:** Calculates the total sum of the values.
* **How it works:** When applied to a DataFrame, it calculates the sum for *each column* individually. If applied to a single Series (one column), it returns a single total number.
* **Common Use Case:** Finding total annual sales, calculating total inventory volume, or summing up user activity points.

```python
data = {
    "Age" : [10,20,30,40],
    "Salary" : [10000,20000,30000,40000]
}

df = pd.DataFrame(data)
print(f"Sum : \n{df.sum()}")  #Sum of Each numaric values
# df.sum()
```

## **2. The `.mean()` Method**
* **What it does:** Calculates the arithmetic average of the values.
* **How it works:** Just like `.sum()`, it automatically scans your numeric columns and returns the average for each one. Missing data (`NaN`) is automatically ignored by default.
* **Common Use Case:** Finding the average age of your user base, calculating average monthly revenue, or finding typical test scores.

```python
data = {
    "Age" : [10,20,30,40],
    "Salary" : [10000,20000,30000,40000]
}

df = pd.DataFrame(data)  
print(f"Mean : \n{df.mean()}")
# df.mean()
```

## **3. The `.describe()` Method (The Ultimate Summary)**
* **What it does:** Generates a comprehensive, high-level statistical summary table of all numeric columns in your DataFrame.
* **What it includes:** In one single command, it returns:
  * **`count`**: Number of non-empty rows (great for spotting missing data).
  * **`mean`**: The average.
  * **`std`**: Standard deviation (how spread out or volatile the data is). (සම්මත අපගමනය)
  * **`min` & `max`**: The absolute lowest and highest values.
  * **`25%, 50%, 75%`**: The percentiles (note: the `50%` is your median). (චතුර්තක -> Q1,Q2,Q2)
* **When to use it:** This should almost always be the very first command you run when you load a brand new dataset to check for extreme outliers and understand your data's distribution.

```python
data = {
    "Age" : [10,20,30,40],
    "Salary" : [10000,20000,30000,40000]
}

df = pd.DataFrame(data)
# print(f"Describe : \n{df.describe()}")
df.describe()
```

## **Drop a Row or Column**

- The code `DataFrame.drop(labels, axis=1, inplace=False)` is used in the Python pandas library to remove one or more columns from a dataset. It generates a brand-new dataset without the chosen columns, leaving your original data completely unchanged.

    - labels: The name (or list of names) of the column(s) you want to remove.
        - Example: `"Age"` to drop one column, or `["Age", "Salary"]` to drop multiple columns.

    - axis=1: Tells pandas to look for the labels horizontally across the columns.By default, pandas looks at `axis=0` **(rows)**. Setting `axis=1` **(or axis='columns')** is required to avoid a "label not found" error when deleting columns.

    - inplace=False: Ensures the method returns a new DataFrame with the specified columns removed.Because it defaults to False, your original DataFrame remains intact. If you want to modify your original dataset directly without assigning it to a new variable, you would change this to inplace=True



```python
data = {
    "Age" : [35,20,22,40],
    "Salary" : [50000,60000,70000,80000],
    "Distance" : [10,20,30,10]
}

df = pd.DataFrame(data)

drop_column = df.drop("Distance", axis=1)
drop_column
```
