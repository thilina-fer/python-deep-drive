# **Logical Errors**

- A logical error in Python occurs when your code runs completely smoothly without crashing or throwing any error messages, but it produces an incorrect or unexpected output. These are often called "bugs" and happen because the algorithm or logic you wrote doesn't match your actual intention.
- Because the Python interpreter cannot detect these flaws automatically, they are considered the hardest type of programming errors to find and fix.

### **Common Types of Logical Errors**

#### **1. Mathematical Operator Precedence**
- Python follows standard mathematical operator precedence (PEMDAS). Forgetting this can skew calculations

```python
# Intending to average 10 and 20 (Expected: 15.0)
result = 10 + 20 / 2 
print(result)  # Outputs: 20.0
```

```python
# he Fix: Wrap the addition in parentheses to force it to evaluate first.
result = (10 + 20) / 2
```

#### **2. Off-by-One Flaws**
- These happen when a loop repeats one time too many or one time too few, often caused by how the range() function works

```python
# Intending to print numbers 1 through 10
for i in range(1, 10):
    print(i)  # Only prints 1 to 9

# The Fix: Increase the upper bound by 1 since range() stops right before the final number.
for i in range(1, 11):
    print(i)  # Now prints 1 to 10
```

#### **3. Mutable Default Arguments**

- Using a mutable object (like a list or dictionary) as a default argument in a function creates a persistent object shared across multiple function calls.

```python
def add_item(item, my_list=[]):
    my_list.append(item)
    return my_list

print(add_item("apple"))  # ['apple']
print(add_item("banana")) # ['apple', 'banana'] (Unexpected!)
```

```python
# The Fix: Use None as the default value instead
def add_item(item, my_list=None):
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list
```

#### **4. Wrong Boolean Operators**
- Mixing up and and or logical operators can make conditional statements execute under the wrong circumstances.

```python
score = 105
# Intending to validate if a score is out of the 0-100 range
if score < 0 and score > 100:
    print("Invalid score")  # Will never print!
```

```python
# The Fix: Change the operator to or because a number cannot simultaneously be below 0 and above 100.
if score < 0 or score > 100:
    print("Invalid score")  # Now correctly identifies invalid scores
```

## **5. Infinite Loops**
- This happens when the loop's termination condition is never satisfied, causing the script to execute indefinitely until stopped manually.

```python
i = 1
while i <= 5:
    print(i)
    i - 1  # Logic error: doesn't modify 'i' in place; should be i -= 1
```
