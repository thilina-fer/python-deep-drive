## **How to Start the Debugger**

You can trigger `pdb` in three primary ways:

* **The Modern Way (Python 3.7+):** Drop `breakpoint()` directly into your source code where you want to pause. This automatically hooks into `pdb` by default.
* **The Legacy Inline Way:** Drop `import pdb; pdb.set_trace()` manually into your code.
* **From the Command Line:** Run your script module directly without modifying the source code by executing `python -m pdb your_script.py`. This drops you into the debugger right at the first line of the file.


## **Core pdb Navigation Commands**

Once execution hits a breakpoint, your terminal changes to an interactive `(Pdb)` prompt. You can use these short command flags to step through the program:

* `n` (next): Executes the current line and advances to the next line in the current function.
* `s` (step): Steps inside the function call on the current line.
* `c` (continue): Resumes standard program execution until the next breakpoint is encountered.
* `r` (return): Continues execution until the current function returns.
* `l` (list): Shows the 11 lines of source code surrounding your current active line.
* `w` (where): Prints the full stack trace, showing the path of execution to the current line.
* `q` (quit): Immediately terminates the debugger and exits script execution.


## **Inspecting and Modifying State**

You can safely view or change your program state dynamically:

* `p <expression>`: Evaluates and prints the value of a variable or an expression (e.g., `p user.name`).
* `pp <expression>`: Pretty-prints complex objects, dictionaries, or data structures for cleaner reading.
* `whatis <variable>`: Prints the exact data type of the designated variable.
* `!<statement>`: Prefixing your input with an exclamation point lets you execute arbitrary Python statements to reassign variables or run hot-fixes on the fly (e.g., `!count = 10`).


## **Breakpoint Management**

* `b <line_number>`: Dynamically adds a new breakpoint at the specified line number of the active file.
* `b`: Run without arguments to list all active breakpoints with their assigned ID numbers.
* `clear <id>`: Permanently deletes the breakpoint corresponding to that specific ID number.

```python
import pdb

def multiplication(a, b):
    answer = a * b
    return answer

pdb.set_trace()
x = input("Enter first number: ")
y = input("Enter second number: ")
mul = multiplication(x, y)
print(mul)
```
