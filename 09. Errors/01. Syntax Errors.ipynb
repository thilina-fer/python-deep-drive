## **01. Syntax Errors**

- A SyntaxError in Python occurs when the interpreter parses your code and encounters a line that violates the structural or grammatical rules of the language. Because Python checks the entire file for syntax validity before running a single instruction, no lines of code will execute if a single syntax error is present

### **Common Causes & Examples**

1. Missing Colons (:)Python requires a colon at the end of control flow statements and declarations, such as if, for, while, def, and class.
- ❌ Wrong: if x > 5 print(x)
**Fix: if x > 5:**

2. Incorrect IndentationPython uses whitespace to define blocks of code. Mixing tabs and spaces, or failing to indent a nested block, triggers an IndentationError (a specific type of syntax error).
- ❌ Wrong:
```
def greet():
print("Hello")  # Error: Expected an indented block
```
**Fix: Indent the line by 4 spaces**.

3. Unmatched Parentheses, Brackets, or QuotesOpening a bracket or quotation mark requires a corresponding closing pair.
- ❌ Wrong: print("Hello" or items = [1, 2, 3
- ❌ Wrong Mixed Quotes: print('Hello")
**Fix: Ensure every (, [, {, ', and " has an exact matching counterpart.**

4. Misusing the Assignment Operator (=)Using a single equals sign (=) inside a conditional statement instead of the comparison operator (==) is invalid syntax.
- ❌ Wrong: if x = 5:
**Fix: if x == 5:**

5. Invalid Variable NamesVariable names cannot begin with numbers or use reserved keywords.
- ❌ Wrong: 3_items = "box" or pass = True
**Fix: Start with a letter or underscore, and avoid keywords like pass, def, or class**
