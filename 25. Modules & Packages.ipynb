# **Modules**

In Python, a module is simply a file containing Python code (functions, classes, or variables) that you can import into other scripts. Think of it as a "toolbox"—instead of writing everything from scratch, you grab a specialized box that already has the tools you need.

### Types of Python Modules

| Category | Description | Examples |
| :--- | :--- | :--- |
| **Built-in Modules** | Part of the Standard Library (pre-installed). | `os`, `sys`, `math`, `datetime`, `random` |
| **Third-Party Modules** | Installed via `pip` from the Python Package Index (PyPI). | `numpy`, `pandas`, `requests`, `matplotlib` |
| **Custom Modules** | Files you create yourself (e.g., `my_logic.py`). | `utils.py`, `database_config.py` |
