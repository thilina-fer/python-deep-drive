# The `while` Loop

A `while` loop repeatedly executes a target statement as long as a given condition is true.

**Basic Syntax:**
```python
while condition:
    # code block to be executed

## For `loop` vs `While` loop

### Comparison: `for` Loop vs. `while` Loop

| Feature | `for` Loop | `while` Loop |
| :--- | :--- | :--- |
| **Usage** | Iterating over a sequence (List, Tuple, Range). | Iterating based on a specific condition. |
| **Length** | Known beforehand (Definite iteration). | Not known beforehand (Indefinite iteration). |
| **Control** | Controlled by the items in the sequence. | Controlled by a boolean expression (True/False). |

Write a python program to read a positive integer as a user input use a while loop to print all numbers from 1 to that input number which are deviciible by 13

````python
num = int(input("Enter a positive integer: "))
i = 13
while i <= num:
    if i % 13 == 0:
        print(i)
    i += 1  
````

Take an integer input and the reverse it.

````python
num = int(input("Enter a number: "))
reverse = 0
while num > 0:
    digit = num%10
    reverse = reverse * 10 + digit
    num = num // 10
print("The reverse of the number is:", reverse)  
````

# The `while-else` Statement

In Python, a `while` loop can have an optional `else` block. 

* **The `else` block runs**: When the `while` condition becomes `False` and the loop finishes naturally.
* **The `else` block does NOT run**: If the loop is stopped prematurely by a `break` statement.

**Syntax:**
```python
while condition:
    # code block
else:
    # runs only if break was not hit

````python
count = 0
while count < 5:
    if count == 2:
        print("Breaking the loop now!")
        break
    print(f"Count is {count}")
    count += 1
else:
    print("This will not print because of the break.")
````
