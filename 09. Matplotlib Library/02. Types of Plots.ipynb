# **Advanced Graphs**

## **01. Histogram**

A histogram is a highly effective visualization tool used to represent the frequency distribution of continuous numerical data.

## **What is a Histogram and Why Use It?**

Unlike a bar chart, which compares distinct categorical variables (like "Apples" vs. "Bananas"), a histogram groups a single continuous range of numbers into equal, consecutive intervals called **bins**. The height of each bar represents how many data points fall into that specific bin.

### **Why we use histograms:**
* **Discovering Data Shape:** Quickly identify if your data follows a normal distribution (a bell curve), is skewed to one side, or is bimodal (has two peaks).
* **Identifying Outliers:** Easily spot data points that fall far outside the normal clustering of your dataset.
* **Understanding Density:** See where the majority of your data is concentrated.

````python
import numpy as np
````

````python
data = np.random.randn(1000)  # Generate 1000 random numbers from a normal distribution
plt.hist(data)

plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
````

### **Customized histogram**

````python
data = np.random.randn(1000)

# Customize bins to 30, change color to green, and add black borders
plt.hist(data, bins=30, color="green", edgecolor="black")

plt.title("Customized Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
````

### **Comparing Multiple Histograms**
Histograms are excellent for comparing two different datasets. When plotting multiple histograms on the same axes, you should use the `alpha` parameter to make the bars semi-transparent so they do not hide each other.

````python
# Generate two different sets of data
data1 = np.random.normal(0, 1, 1000)   # Center at 0
data2 = np.random.normal(2, 1.5, 1000) # Center at 2, wider spread

# Plot both histograms with transparency (alpha)
plt.hist(data1, bins=30, alpha=0.5, label="Dataset A", edgecolor="black")
plt.hist(data2, bins=30, alpha=0.5, label="Dataset B", edgecolor="black")

plt.title("Overlapping Histograms")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend(loc="upper right")

plt.show()
````

### **Advanced: Normalizing to Density**
Sometimes you want the y-axis to represent a probability density rather than a raw count (meaning the total area under the histogram equals 1). This is useful when comparing datasets of different sizes.

You can achieve this by setting density=True.

````python
data = np.random.randn(1000)

# Set density=True to show probability density instead of raw counts
plt.hist(data, bins=30, density=True, edgecolor="black", color="purple")

plt.title("Density Histogram")
plt.xlabel("Value")
plt.ylabel("Probability Density")

plt.show()
````

## **02. Box Plots**

A box plot (or box-and-whisker plot) is a standardized way of displaying the distribution of data based on a five-number summary. 

## What is a Box Plot and Why Use It?

While a histogram shows you the shape and density of your data, a box plot abstracts the data into specific statistical markers.

### The Five-Number Summary:
1. **Minimum:** The lowest data point (excluding extreme outliers).
2. **First Quartile (Q1):** The 25th percentile (the bottom of the box).
3. **Median (Q2):** The 50th percentile (the line inside the box).
4. **Third Quartile (Q3):** The 75th percentile (the top of the box).
5. **Maximum:** The highest data point (excluding extreme outliers).

*Note: The distance between Q1 and Q3 is called the Interquartile Range (IQR). The "whiskers" extend to show the rest of the distribution, while points outside the whiskers are plotted individually as **outliers**.*

### Why we use box plots:
* **Outlier Detection:** They are the absolute best visualization for quickly spotting outliers in your data.
* **Comparing Distributions:** They allow you to easily compare the medians, spreads, and skews of multiple datasets side-by-side without the plot becoming cluttered.

````python
data = [20, 25, 27, 29, 30, 32, 35, 50, 100] 

plt.boxplot(data)
plt.title("Box Plot Example")
plt.show()
````

````python
# Generate random data with a normal distribution
# (mean=100, standard deviation=10, 200 data points)
data = np.random.normal(100, 10, 200)

# Add a few extreme outliers manually to see how they render
data = np.append(data, [140, 150, 50, 45])

plt.boxplot(data)

plt.title("Basic Box Plot")
plt.ylabel("Values")

plt.show()
````

### **Comparing Multiple Datasets**

Box plots truly shine when you place them side-by-side. To do this, pass a list of datasets to the boxplot() function instead of a single array.

````python
# Generate three different datasets
data_A = np.random.normal(100, 10, 200)
data_B = np.random.normal(90, 20, 200)  # Wider spread
data_C = np.random.normal(110, 5, 200)  # Tighter spread, higher median

# Group them into a list
data_to_plot = [data_A, data_B, data_C]

plt.boxplot(data_to_plot, label=["Group A", "Group B", "Group C"])

plt.title("Comparing Multiple Distributions")
plt.ylabel("Scores")

plt.show()
````

### **Customizing Box Plots**

You can change the orientation, add visual styling, and alter how outliers are displayed.

- Key Customization Parameters:
    - vert=False: Draws the boxes horizontally instead of vertically.

    - notch=True: Creates a V-shaped notch around the median to represent confidence intervals.

    - patch_artist=True: Allows the boxes to be filled with color (by default, they are just outlines).

````python
data = np.random.normal(100, 10, 200)

# Create a horizontal, filled, notched box plot
box = plt.boxplot(data, vert=False, notch=True, patch_artist=True)

# Customize the color of the box
for patch in box['boxes']:
    patch.set_facecolor('lightblue')
    
# Customize the median line
for median in box['medians']:
    median.set_color('red')
    median.set_linewidth(2)

plt.title("Customized Horizontal Box Plot")
plt.xlabel("Values")

plt.show()
````

## **03. Stack Plot (Area Plot)**

A stack plot (or stacked area plot) is used to show the breakdown of a total amount over time or across categories. It visualizes how different parts contribute to a whole, and how that whole changes.

### **What is a Stack Plot and Why Use It?**

Imagine a line chart with multiple lines representing different categories. A stack plot takes those lines and "stacks" them on top of one another, filling in the area underneath. 

### **Why we use stack plots:**
* **Tracking Totals:** The top edge of the entire plotted area represents the grand total of all categories combined.
* **Showing Proportions over Time:** You can easily see how much a specific category contributes to the total at any given point on the x-axis.
* **Visualizing Cumulative Trends:** Excellent for resource tracking, budget breakdowns, or measuring team outputs over days, months, or years.

*Note: Stack plots are best used when you want to emphasize the total and the general trends of the components. They are not ideal for precisely comparing the exact values of the individual stacked layers, as their baselines fluctuate.*


## 1. Creating a Basic Stack Plot

To create a stack plot, use the `plt.stackplot()` function. You need one array for the x-axis (usually time) and multiple arrays for the y-axis (the categories you are stacking).

```python
import matplotlib.pyplot as plt

# Define the x-axis (e.g., days of the week)
days = [1, 2, 3, 4, 5]

# Define the y-axis data for three different categories (e.g., hours spent on tasks)
coding = [3, 4, 2, 5, 4]
meetings = [1, 2, 3, 1, 2]
documentation = [2, 1, 1, 2, 3]

# Create the stack plot
plt.stackplot(days, coding, meetings, documentation)

plt.title("Basic Stack Plot")
plt.xlabel("Days")
plt.ylabel("Hours Spent")

plt.show()

````python
# Define the x-axis (e.g., days of the week
days = [1,2,3,4,5,6,7]

# Define the y-axis data for three different categories (e.g., hours spent on tasks)
codings= [2, 3, 5, 7, 11, 13, 17]
meetings= [1, 2, 3, 4, 5, 6, 7]
documentation= [1, 1, 2, 3, 5, 8, 13]

# Create the stack plot
plt.stackplot(days, codings, meetings, documentation, labels=['Coding', 'Meetings', 'Documents'], alpha=0.8)

plt.title("Stacked Area Chart Example")
plt.xlabel("Days")
plt.ylabel("Hours Spent on Activities")

plt.show()
````

## **Adding Labels and Customizing Colors**

A stack plot is unreadable without a legend indicating what each layer represents. You can pass a list of labels and a list of colors directly into the `stackplot()` function.

````python
# Define the x-axis (e.g., days of the week
days = [1,2,3,4,5,6,7]

# Define the y-axis data for three different categories (e.g., hours spent on tasks)
codings= [2, 3, 5, 7, 11, 13, 17]
meetings= [1, 2, 3, 4, 5, 6, 7]
documentation= [1, 1, 2, 3, 5, 8, 13]

# Custom colors for each category

colors = ["#F90000", "#0080FF", "#00FF00"]

# Create the stack plot
# alpha=0.8 for transparency
plt.stackplot(days, codings, meetings, documentation, labels=['Coding', 'Meetings', 'Documents'], alpha=0.8, colors=colors)

plt.title("Stacked Area Chart Example")
plt.xlabel("Days")
plt.ylabel("Hours Spent on Activities")

plt.legend(loc='upper left')  # Add a legend to identify each category

plt.show()
````

## **Alternative Approach: Plotting a Dictionary**

If your data is stored in a dictionary (which is common when working with data processing libraries like Pandas), you can use the .`values()` method to plot it cleanly.

````python
# X-axis data
years = [2018, 2019, 2020, 2021, 2022]

# Y-axis data stored in a dictionary
revenue = {
    "Product A": [10, 15, 20, 22, 25],
    "Product B": [5, 8, 12, 18, 24],
    "Services": [15, 12, 10, 8, 5]
}

# Plot using dictionary values and keys
plt.stackplot(years, revenue.values(), labels=revenue.keys(), alpha=0.8)

plt.legend(loc="upper left")
plt.title("Company Revenue Streams")
plt.xlabel("Year")
plt.ylabel("Revenue (in Millions)")

plt.show()
````

## **Best Practices for Stack Plots**

- **Limit the Categories**: Try not to stack more than 4 or 5 categories. Too many layers become visual clutter and are impossible to read.

- **Order Matters**: Place the largest, most stable, or most important data category at the bottom. The bottom layer is the only one with a flat zero baseline, making it the easiest to read accurately.

- **Use Transparency (alpha)**: Setting an alpha value (between 0 and 1) can soften the colors and make the plot easier on the eyes, especially if you have grid lines turned on.
