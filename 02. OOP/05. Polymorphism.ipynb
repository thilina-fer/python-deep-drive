# **Polymorphism**

**Polymorphism** comes from two Greek words: **"Poly"** (Many) and **"Morph"** (Forms). In programming, it means **"Many Forms"**. 

It allows different classes to have methods with the **exact same name** but **different behaviors**. 

### Real-World Analogy:
Think of the **"Start"** action on different vehicles.
* If you "Start" a **Car**, you turn a key or press a button.
* If you "Start" a **Motorcycle**, you might use a kick-starter.
* If you "Start" an **Airplane**, it requires a complex cockpit check and propeller rotation.

The action name is exactly the same (**"Start"**), but the way each vehicle does it is completely different. This is Polymorphism!

### Why is it useful?
It enables you to write flexible and reusable code. You can call the same method name on a list of different objects without worrying about their specific types.

```python
# 1. Parent Class (Superclass)
class Vehicle:
    def start(self):
        # Default behavior for a general vehicle
        print("Starting a generic vehicle...")


# 2. Child Class 1 (Subclass)
class Car(Vehicle):
    # METHOD OVERRIDING: Customizing the 'start' method for a Car
    def start(self):
        print("Car starts: Turning the ignition key or pressing the Start button.")


# 3. Child Class 2 (Subclass)
class Motorcycle(Vehicle):
    # METHOD OVERRIDING: Customizing the 'start' method for a Motorcycle
    def start(self):
        print("Motorcycle starts: Using the kick-starter or handlebar switch.")


# --- Testing Polymorphism (Many Forms) ---

# Creating a list of different vehicle objects
# Both Car and Motorcycle are treated as forms of a common Superclass (Vehicle)
garage = [Car(), Motorcycle()]

print("--- Calling the SAME method name on DIFFERENT objects ---")
for vehicle in garage:
    # Python automatically calls the correct 'start' method for each object!
    vehicle.start()
```

### Comparison: Polymorphism in Java vs Python

| Feature | Java ☕ | Python 🐍 |
| :--- | :--- | :--- |
| **Compile-time Polymorphism** <br>*(Method Overloading)* | **Supported** <br>Multiple methods can have the same name if their parameters are different. | **Not Directly Supported** <br>Writing a new method with the same name will overwrite the older one. (Achieved via Default Arguments). |
| **Runtime Polymorphism** <br>*(Method Overriding)* | **Supported** <br>Child class overrides the parent class method; resolved at runtime based on object type. | **Fully Supported** <br>This is the primary way polymorphism works in Python (along with Duck Typing). |
| **Language Nature** | **Statically-typed** <br>Data types are checked before running the code (at compile time). | **Dynamically-typed** <br>Data types are checked while the code is running (at runtime). |
