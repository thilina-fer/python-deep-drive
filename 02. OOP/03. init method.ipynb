# **What is the `__init__` Method in Python? (The Constructor)**

The **`__init__`** method is a special function in Python classes. It is also known as a **Constructor**. 

Whenever you create a new object from a class, Python automatically calls the `__init__` method **first** behind the scenes.

* **The Name:** It has two underscores before and after the word "init" (`__init__`).
* **The Purpose:** Its main job is to **initialize** (set up) the initial data and attributes of a new object right at the moment it is born.

### Real-World Analogy:
Imagine an **Identity Card (ID Card) factory**. 
* The blueprint of the ID card specifies that every card must have a **Name**, **Age**, and **ID Number**.
* When the machine prints a new card for a specific person, it immediately stamps their real name and age onto that card.
* The `__init__` method is like that stamping machine. It sets up the unique details of the object as soon as it is created.

```python
# Defining the Class (The Blueprint)
class Student:
    # The Constructor method - Runs automatically when an object is made
    def __init__(self, full_name, student_age):
        print(f"--- Creating a new student profile for {full_name} ---")
        
        # Storing the passed data inside the object using 'self'
        self.name = full_name
        self.age = student_age

    # A regular method to display information
    def display_info(self):
        print(f"Student Name: {self.name} | Age: {self.age}")


# --- Creating Objects and passing data directly ---

# Notice how we pass "Nimal" and 21 inside the brackets Student()
# This data goes straight to the __init__ method
student1 = Student("Nimal", 21)
student2 = Student("Kamal", 23)

print("\n--- Displaying Profiles ---")
# Calling the normal method to check data
student1.display_info()  # Output: Student Name: Nimal | Age: 21
student2.display_info()  # Output: Student Name: Kamal | Age: 23
```
