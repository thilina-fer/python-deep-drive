# **Inheritance**

**Inheritance** is a core pillar of Object-Oriented Programming (OOP). It allows a new class (called **Child / Sub / Derived Class**) to inherit properties and methods from an existing class (called **Parent / Super / Base Class**).

* **Main Goal:** To promote **Code Reuse** and build a clean relationship between classes.

### Real-World Analogy:
Think of a **Vehicle** blueprint as the Parent Class. It has common features like `brand`, `model`, and a function called `start_engine()`. 
* A **Car** and a **Motorcycle** are Child Classes. 
* They automatically get the `brand` and `start_engine()` function from the Vehicle class without rewriting them.
* However, a Car can add its own unique feature like `number_of_doors`, and a Motorcycle can add `has_sidecar`.

---

### Additional Advanced Facts about Inheritance

Here are some important concepts you must know for interviews and practical coding:

#### 1. Types of Inheritance in Python
Python is very flexible and supports multiple types of inheritance:
* **Single Inheritance:** A child class inherits from only **one** parent class (e.g., `Dog` inherits from `Animal`).
* **Multiple Inheritance:** A child class can inherit from **more than one** parent class at the same time. *(Note: Languages like Java do not support this, but Python does!)*
* **Multilevel Inheritance:** A child class inherits from a parent class, which also inherits from another parent class (Grandparent -> Parent -> Child).

#### 2. Method Overriding
* If a child class wants to change the behavior of a method it got from the parent class, it can redefine that method with the exact same name.
* **Example:** The parent class `Animal` has a method `speak()` that does nothing. The child class `Dog` overrides `speak()` to print "Woof!".

#### 3. The `super()` Function
* The `super()` function is a built-in shortcut that lets the child class call methods (especially the `__init__` constructor) of the parent class.
* This ensures that the parent class is initialized properly before the child class adds its own data.

```python
# 1. Parent Class (Superclass / Base Class)
class Animal:
    def __init__(self, name):
        # Attribute shared by all animals
        self.name = name  

    def eat(self):
        # Method shared by all animals
        print(f"{self.name} is eating food.")


# 2. Child Class (Subclass / Derived Class)
# By passing 'Animal' inside the brackets, Dog inherits everything from Animal
class Dog(Animal):  
    def __init__(self, name, breed):
        # Calling the Parent class constructor to initialize the name
        super().__init__(name)

        # self.name = name  # Initializing the inherited attribute 'name' directly  
          
        # Adding a unique attribute specific only to Dogs
        self.breed = breed  

    # METHOD OVERRIDING: Customizing the speak method for Dog
    def speak(self):
        return f"{self.name} says Woof! Woof!" 


# 3. Another Child Class (Subclass / Derived Class)
# Cat also inherits everything from Animal
class Cat(Animal):  
    # METHOD OVERRIDING: Customizing the speak method for Cat
    def speak(self):
        return f"{self.name} says Meow!"  


# --- Testing the Inheritance ---

print("--- Creating Objects ---")
# Creating instances of the Child classes
my_dog = Dog("Buddy", "Golden Retriever")  
my_cat = Cat("Whiskers")  

print("\n--- Calling Inherited Methods (From Parent Class) ---")
# The 'eat' method was written only in Animal, but both Dog and Cat can use it
my_dog.eat()
my_cat.eat()

print("\n--- Calling Overridden & Unique Methods (From Child Classes) ---")
# Calling the customized 'speak' methods
print(my_dog.speak())  # Output: Buddy says Woof! Woof! [cite: 190]
print(my_cat.speak())  # Output: Whiskers says Meow! [cite: 193]

print("\n--- Accessing Attributes ---")
# Accessing the parent class attribute and child class attribute
print(f"Dog's Name: {my_dog.name}")  # Output: Dog's Name: Buddy [cite: 188, 189]
print(f"Dog's Breed: {my_dog.breed}")  # Output: Dog's Breed: Golden Retriever
```
