# **Encapsulation**

**Encapsulation** is one of the fundamental pillars of Object-Oriented Programming (OOP).It means **bundling data (attributes) and the methods (functions)** that work on that data into a single unit a **Class**. 

At the same time, it restricts direct access to some of the object's components, which is known as **Data Hiding**

### Real-World Analogy:
Think of a **Medical Capsule (Pill)**. 
* Inside the capsule, there is a specific mix of medicines (Data and Internal Methods).
* The outer plastic shell protects the medicine from the outside environment (Encapsulation)
* You cannot touch or change the medicine powder directly. You can only swallow the pill (using a Public Method) to let it work safely.

---

### 💡 Benefits of Encapsulation
* **Data Hiding:** Users of the class only see *what* the object does, not *how* it does it internally
* **Controlled Access:** Prevents accidental modification of data by outside code.
* **Flexibility & Modularity:** You can change the internal implementation of a class without breaking the code that uses it

---

### 🔒 Access Levels (Modifiers) in Python
Unlike languages like Java or C++, Python does not have strict keywords like `public` or `private`. Instead, Python uses **Naming Conventions** (underscores) to define access levels:

1. **Public Attributes:** No underscores (e.g., `self.name`).They can be accessed and modified from anywhere outside the class.
2. **Protected Attributes:** Single underscore (e.g., `self._balance`).This is a hint to other programmers that the variable is intended only for internal use and subclasses. Python does not strictly enforce this.
3. **Private Attributes:** Double underscores (e.g., `self.__account_number`). They **cannot** be accessed or modified directly from outside the class.

---

### 🛠️ Getter and Setter Methods
To access or change private attributes safely, we write special public functions inside the class:
* **Getter Method:** Allows outside code to safely **read** a private value.
* **Setter Method:** Allows outside code to safely **modify** a private value after checking if the new data is valid.

## **Example: Encapsulation in BankAccount Class**

This notebook cell explains how **Encapsulation (Data Hiding)** is practically implemented in the `BankAccount` Python code snippet.


### 1. Public vs Private Attributes (Data Hiding)
* **`self.holder_name` (Public):** No underscores are used. Anyone can read or change the account holder's name directly from outside the class using `account.holder_name`.
* **`self.__account_number` & `self.__balance` (Private):** Written with **double underscores (`__`)**. They are completely hidden from the outside world. Trying to access them directly (e.g., `account.__balance`) will throw an `AttributeError`.


### 2. Private Methods
* **`def __validate_amount(self, amount):`** This is a hidden internal helper function. 
* It is only used *inside* the class to check if an amount is greater than zero. Outside code cannot call this function directly.

### 3. Accessing Data Safely (Getter Methods)
Since the balance and account number are private, we use **Getter methods** to safely read their values:
* `get_balance()` -> Returns the private balance safely.
* `get_account_number()` -> Returns the private account number safely.


### 4. Modifying Data Safely (Public Methods as Setters)
Instead of letting outside code change the balance directly, the class provides controlled public methods:
* **`deposit(amount)`:** Validates the amount first using the private method, then adds it to the balance.
* **`withdraw(amount)`:** Checks if the amount is valid and if there is enough money in the account before deducting it. This prevents errors like ending up with a negative bank balance.


### 📥 Code Execution Output Summary

When you run the code, it behaves exactly as intended by OOP encapsulation standards:
1. **Allowed:** Reading the public holder's name.
2. **Allowed:** Reading balance and account numbers using safe `.get_balance()` and `.get_account_number()` methods.
3. **Allowed:** Modifying balance safely via `.deposit()` and `.withdraw()`.
4. **Blocked (Error Caught):** Trying to bypass security to access `__balance` or `__validate_amount` directly is strictly blocked by Python.

```python
# Defining the Class with Encapsulation
class BankAccount:
    def __init__(self, account_holder, account_number, initial_balance):
        # 1. PUBLIC ATTRIBUTE: Accessible by anyone
        self.holder_name = account_holder 
        
        # 2. PRIVATE ATTRIBUTES: Hidden from the outside world
        self.__account_number = account_number 
        self.__balance = initial_balance 

    # 3. PRIVATE METHOD: Can only be called inside this class
    def __validate_amount(self, amount):
        return amount > 0

    # 4. GETTER METHOD: Safely exposes the private balance
    def get_balance(self):
        return self.__balance

    # 5. GETTER METHOD: Safely exposes the private account number
    def get_account_number(self):
        return self.__account_number

    # 6. PUBLIC METHOD (Acts as a Setter): Controls modification of balance
    def deposit(self, amount):
        if self.__validate_amount(amount):
            self.__balance += amount
            print(f"Deposited: ${amount} | New Balance: ${self.__balance}")
        else:
            print("Invalid deposit amount!")

    # 7. PUBLIC METHOD (Acts as a Setter): Controls modification of balance
    def withdraw(self, amount):
        if self.__validate_amount(amount) and amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: ${amount} | Remaining Balance: ${self.__balance}")
        else:
            print("Transaction Denied: Invalid amount or insufficient funds!")


# --- Testing Encapsulation ---

print("--- Creating Bank Account Object ---")
account = BankAccount("John Doe", "987654321", 1000.0)

print("\n--- Accessing Public Data ---")
# This works perfectly because holder_name is public
print(f"Account Holder: {account.holder_name}") 

print("\n--- Accessing Private Data via Getters ---")
# We cannot do account.__balance. We MUST use the getter methods
print(f"Account Number: {account.get_account_number()}")
print(f"Starting Balance: ${account.get_balance()}")

print("\n--- Modifying Private Data via Public Methods ---")
# Safely depositing and withdrawing money
account.deposit(500.0)
account.withdraw(200.0)

print("\n--- Attempting to break Encapsulation (Will Fail) ---")
try:
    # Trying to access a private variable directly throws an Error
    print(account.__balance)
except AttributeError as e:
    print(f"Error caught successfully: {e}")

try:
    # Trying to call a private validation method directly throws an Error
    account.__validate_amount(100)
except AttributeError as e:
    print(f"Error caught successfully: {e}")
```

## **Example 2: User Profile Management System**

In this example, we model a **UserProfile** class. 
* A user's **Username** is public because everyone can see it.
* However, a user's **Password** must be **Private** (`__password`) so that it cannot be read or changed directly from outside code.
* We use a **Setter Method** (`change_password`) to allow password changes, but ONLY if the user provides the correct old password first!

```python
class UserProfile:
    def __init__(self, username, email, initial_password):
        # PUBLIC ATTRIBUTES: Anyone can view or change these
        self.username = username
        self.email = email
        
        # PRIVATE ATTRIBUTE: Hidden from the outside code for security
        self.__password = initial_password

    # GETTER METHOD: Safely checks if the login password is correct
    # (It does not show the password, it only validates it)
    def check_login(self, input_password):
        return self.__password == input_password

    # SETTER METHOD: Controls how the private password is modified
    def change_password(self, old_password, new_password):
        # Step 1: Verification (Security Check)
        if self.__password == old_password:
            # Step 2: Validation (Data Quality Check)
            if len(new_password) >= 6:
                self.__password = new_password
                print(f"Success: Password for '{self.username}' has been updated.")
            else:
                print("Error: New password must be at least 6 characters long!")
        else:
            print("Error: Authentication Failed! Incorrect old password.")


# --- Testing the User Profile System ---

print("--- Creating User Profile ---")
user = UserProfile("asanka_ai", "asanka@email.com", "Secret123")

print("\n--- Accessing Public Data ---")
# Reading and modifying public data works fine
print(f"Username: {user.username}")
user.email = "asanka_new@email.com"
print(f"Updated Email: {user.email}")

print("\n--- Trying to Login ---")
# Using the getter/validator method
is_success = user.check_login("WrongPass")
print(f"Login with 'WrongPass': {is_success}") # Output: False

is_success = user.check_login("Secret123")
print(f"Login with 'Secret123': {is_success}") # Output: True

print("\n--- Attempting to Change Password ---")
# 1. Trying with a wrong old password (Will fail)
user.change_password("WrongOldPass", "NewPassword456")

# 2. Trying with a short new password (Will fail validation)
user.change_password("Secret123", "123")

# 3. Providing correct data (Will succeed)
user.change_password("Secret123", "ValidNewPass99")

print("\n--- Attempting to break Encapsulation (Will Fail) ---")
try:
    # Trying to print or hack the private password directly throws an Error
    print(user.__password)
except AttributeError as e:
    print(f"Encapsulation works! Securely blocked direct access: {e}")
```
