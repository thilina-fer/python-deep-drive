# **Beautiful Soup**

Beautiful Soup is a popular Python library used for parsing HTML and XML documents to extract specific data for web scraping. It creates a structured parse tree from raw web page source code, making it easy to navigate, search, and modify elements even if the markup is messy or broken

Installation

```
pip install beautifulsoup4 requests
```

Key Methods & Cheatsheet

### Common BeautifulSoup Methods

| Method / Syntax | Description |
| :--- | :--- |
| `BeautifulSoup(html_content, "html.parser")` | Initializes the parser engine. |
| `soup.find("tag")` | Locates the very first instance of a specific HTML tag. |
| `soup.find_all("tag")` | Returns a list/array containing every matching tag found on the page. |
| `soup.find("tag", class_="name")` | Filters elements by their HTML class name (note the trailing underscore). |
| `soup.find("tag", id="name")` | Filters elements by their unique HTML ID attribute. |
| `element.text` or `element.get_text()` | Extracts only the human-readable text from inside an HTML element, stripping out the tags. |
| `element.get("attribute")` | Extracts the value of a specific tag attribute, such as grabbing a URL from an `href` attribute. |
| `soup.select("css-selector")` | Uses standard CSS selectors (like `.class-name` or `#id-name`) to find components. |


````python
from bs4 import BeautifulSoup
html_doc = """<html><head><title>The Dormouse's story</title></head>
<body>
<p class="title"><b>The Dormouse's story</b></p>
<p class="story">Once upon a time there were three little sisters; and their names were
<a href="http://example.com/elsie&quot; class="sister" id="link1">Elsie</a>,
<a href="http://example.com/lacie&quot; class="sister" id="link2">Lacie</a> and
<a href="http://example.com/tillie&quot; class="sister" id="link3">Tillie</a>;
and they lived at the bottom of a well.</p>
<div>This is a Div element</div>
<p class="story">...</p>
"""
````

````python
soup = BeautifulSoup(html_doc,"html.parser")
print(soup.title)
````

````python
print(soup.p)
print(soup.p.text)
````

````python
print(soup.a)
````

````python
print(soup.div)
````

````python
print(soup.find("a"))
````

````python
print(soup.find_all("a"))
````

````python
print(soup.find("p"))
````

````python
print(soup.find_all("p"))
````

**Q.**
Featch a webpage, check if the request was successfull and exact the main heading text from the page's content. You can assume that the base url is `https://example.com/` 

````python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://example.com/")

if response.status_code == 200:
    if "text/html" in response.headers['Content-Type']:
        soup = BeautifulSoup(response.content, "html.parser")
        if soup.h1.text:
            print(soup.title.text)
        else:
            print("The page does not contain an <h1> tag.")
    else:
        print("The content is not HTML.")
else:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")
````
