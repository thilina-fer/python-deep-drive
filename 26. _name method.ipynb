# **if __name__ = "__main__" construct**

- This tells python to only run some code when the file is being run derectly, and not when it's being imported as a module in another file
- This construct allows you to write code that can be both reusable as a module and executable as a standalone program.

- Every Python file has a built-in variable called __name__. Python automatically sets the value of this variable depending on how you are running the file:

    - **Direct Execution** : If you run python myscript.py, Python sets __name__ to "__main__".
    - **Imported Execution** : If you write import myscript in another file, Python sets __name__ to the filename (e.g., "myscript").

```python
def add(a, b):
    return a + b

# This part only runs if you run this specific file
if __name__ == "__main__":
    print("Direct Execution Mode")
    print(add(10, 5))
```

- Scenario A: Direct Run
    - If you run python math_tools.py in your terminal:

    Output:
    
        Direct Execution Mode
        15

- Scenario B: Imported
    - If you create main.py and write import math_tools:

    Output: 

        (Nothing)
        The code inside the if block is ignored, but you can still use the add function.
