# Python Functions and Functional Programming

Functions allow you to write code once and use it multiple times. 

**Key Components:**
* **Declaration**: Use the `def` keyword.
* **Parameters**: Variables listed inside the parentheses in the function definition.
* **Arguments**: The actual values sent to the function when it is called.
* **Return Statement**: Sends a result back to the caller using the `return` keyword.

**Syntax:**
```python
def function_name(parameter):
    # block of code
    return result

## Comparison: Python Functions vs. Java Methods

| Feature | Python | Java |
| :--- | :--- | :--- |
| **Keyword** | `def` | Requires access modifiers (e.g., `public static`) |
| **Data Types** | Dynamic (not required) | Static (must define `String`, `int`, etc.) |
| **Class Requirement** | Can exist alone (Global) | Must be inside a **Class** |
| **Return Type** | Not declared | Must be declared (or use `void`) |
| **Indentation** | Uses indentation to define the block | Uses curly braces `{ }` |

### Python Function Declaration

To declare a function in Python, you follow a specific structure:

1. **`def` Keyword**: Tells Python you are defining a function.
2. **Function Name**: Should be descriptive and follow `snake_case`.
3. **Parameters**: Optional variables inside parentheses `()` that receive data.Do not add a data type on parameters
4. **Colon `:`**: Marks the end of the header.
5. **Docstring**: An optional triple-quoted string `""" """` to describe what the function does.
6. **Function Body**: The indented block of code to be executed.
7. **Return Statement**: An optional `return` to send a result back to the caller.

````python
def sum_of_two_numbers(a,b):
    return a + b
print(sum_of_two_numbers(3,4)) #Output : 7
````

## Parameters vs. Arguments

While they both involve passing data into a function, the distinction lies in **where** they are defined:

*   **Parameters**: The variables listed in the function's definition. They act as placeholders for the data the function expects to receive.
*   **Arguments**: The actual values or data passed to the function when it is called.

| Feature | Parameters | Arguments |
| :--- | :--- | :--- |
| **Location** | Defined in the function signature (`def` line). | Provided in the function call. |
| **Purpose** | To define what kind of data is needed. | To provide the specific data to work with. |
| **Nature** | Variables (names). | Values (data). |

````python
# 'name' and 'age' are PARAMETERS (Placeholders)
def display_info(name, age):
    print(f"Name: {name}, Age: {age}")

# "Kasun" and 25 are ARGUMENTS (Actual Values)
display_info("Kasun", 25)
````
