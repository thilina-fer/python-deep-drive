# **JSON Lines (JSONL) Data Type**

**JSONL (JSON Lines)** is a text-based format where **each individual line is a valid, independent JSON object**. 

Unlike a standard JSON file (which wraps everything inside a single massive array or object), JSONL stores records line-by-line. This makes it the industry standard format for **large-scale datasets, streaming data, and AI/Machine Learning logging** (such as fine-tuning datasets for LLMs) because you can process it line-by-line without loading the entire file into memory at once.


#### 🌐 Standard JSON Array (`dataset.json`):
```json
# Normal JSON
[
    {"id": 1, "name": "Amal"},
    {"id": 2, "name": "Nimal"},
    {"id": 3, "name": "Sunil"}
]


# JSON Lines (dataset.jsonl):
{"id": 1, "name": "Amal"}
{"id": 2, "name": "Nimal"}
{"id": 3, "name": "Sunil"}

````python
import json

with open("data.jsonl","r") as file :
    for line in file:
        if line == "\n":
            continue
        data = json.loads(line)
        print(data.get("name"))
````

````python
import json

with open("data.jsonl","r") as file :
    for line in file:
        if line.strip():
            data = json.loads(line)
            print(data.get("name"))
       
````

## **Comparison: `json.load()` vs. `json.loads()`**

Both `json.load()` and `json.loads()` are used for **Deserialization** (converting text-based JSON data back into a native Python object like a `dict` or `list`). 

The fundamental difference lies in **what kind of input** they accept: one parses data directly from a physical file, while the other parses data from a string variable inside your code.



## Quick Comparison Table

| Feature | `json.load()` (No "s")  | `json.loads()` (Has "s" for **S**tring)  |
| :--- | :--- | :--- |
| **Primary Input** | Accepts an open **File Object** (file stream). | Accepts a raw **String variable** containing JSON data. |
| **Typical Use Case** | Reading data stored inside external `.json` files. | Parsing JSON data received from a web API request. |
| **Required Syntax** | Must be used inside a file block (like `with open()`). | Can be run directly anywhere in your script. |
| **Example Argument** | `json.load(file_pointer)` | `json.loads(json_string_variable)` |
