# **Working with CSV Data in Python**

**CSV (Comma-Separated Values)** is one of the most widely used plain-text formats for storing tabular data (numbers and text). It maps perfectly to spreadsheets (like Microsoft Excel or Google Sheets) and data frames in data science.

Each line of the file is a data record, and each record consists of one or more fields, separated by a **comma (`,`)**.


### Structural Anatomy of a CSV File

A typical CSV file consists of an optional **Header Row** followed by rows of actual data records.

#### Raw Text Format (`students.csv`):
```text
id,name,major,gpa
1,Nimasha Perera,Data Science,3.85
2,Kasun Fernando,Cyber Security,3.20
3,Dilshan Silva,AI Engineering,3.92

````python
import csv

with open("sample.csv", "r", newline="") as file:
    reader_file = csv.reader(file)
    for row in reader_file:
        print(row)
````

````python
import csv

with open("sample.csv","r", newline="") as file:
    reader_file = csv.DictReader(file)
    for row in reader_file:
        print(row)
````

## **Writing CSV Data**

Writing data to a CSV (Comma-Separated Values) file can be done using Python's built-in **`csv`** module. It provides two main approaches: writing data as **Lists** using `csv.writer`, or writing data as **Dictionaries** using `csv.DictWriter`.

````python
csv_file_path = "sample_output.csv"

data = [
    {'name': 'Alice Johnson', 'age': '28', 'city': 'New York'},
    {'name': 'Bob Smith', 'age': '34', 'city': 'Los Angeles'},
    {'name': 'Charlie Brown', 'age': '45', 'city': 'Chicago'},
    {'name': 'Diana Prince', 'age': '29', 'city': 'Houston'},
    {'name': 'Ethan Hunt', 'age': '38', 'city': 'Phoenix'}
]

field_name = ["name", "age", "city" ]

with open(csv_file_path, "w", newline="") as csv_file:
    csv_writer = csv.DictWriter(csv_file, fieldnames=field_name)
    csv_writer.writeheader()
    csv_writer.writerows(data)
````

## Quick Comparison: `writer` vs `DictWriter`

| Feature | `csv.writer` 📝 | `csv.DictWriter` 🗂️ |
| :--- | :--- | :--- |
| **Data Structure** | List of lists / tuples. | List of Python dictionaries. |
| **Header Configuration** | Handled manually as the first array list item. | Defined explicitly using the `fieldnames` parameter. |
| **Methods Used** | `.writerow()` or `.writerows()` | `.writeheader()` followed by `.writerows()` |
| **Order Dependency** | **Strict.** Items must be ordered perfectly to match columns. | **Flexible.** Key names match automatically regardless of dictionary sequence. |

**Q.** 
A school store student examination records in the csv file name students.csv . The file contains the following columns.student_id, name, maths, science, english.
- S001,Kamal,78,50,65 
- S002,Amal,70,40,85 
- S003,Nimal,80,30,95 

- Write a python program using the csv module to :
    1. Read all student records from the csv file
    2. Calculate a total marks and avarage marks of each student
    3. Dittermine wheather each student is pass or Fail(pass mark for each subject is 50)
    4. Display students whos avarage marks are above 75
    5. Save fail student details into another csv file name fail_students.csv

````python
import csv

with open("students.csv","r",newline="") as file:
    reader_file = csv.DictReader(file)
    for row in reader_file:
        print(row)

        # Total Marks
        total = int(row["maths"]) + int(row["science"]) + int(row["english"])
        print(f"Total marks {row['name']}: {total}")

        # Average Marks
        avarage = total / 3
        print(f"Avarage marks {row['name']}: {avarage}") 

    
        pass_fail = int(row["maths"]) >= 50 and int(row["science"]) >= 50 and int(row["english"]) >= 50
        if pass_fail:
            print(f"Pass: {row['name']}\n")
        else:
            print(f"Fail: {row['name']}\n")

        if avarage > 75:
            print(f"Avarage Marks > 75: {row['name']}\n")
            

        field_name = ["student_id", "name", "maths", "science", "english"]
        with open("fail_students.csv", "w", newline="") as fail_file:
            fail_writer = csv.DictWriter(fail_file, fieldnames=field_name)
            fail_writer.writeheader()
            if pass_fail:
                fail_writer.writerow(row)
            
            
````
