# **File Handling in Python**

File handling is a crucial part of Data Science and Software Development. It allows Python to communicate with your computer's storage to **create, read, update, and delete** files (like `.txt`, `.csv`, or `.json`).


### 1. The Core Concept: The `open()` Function

To work with any file, you must use the built-in `open()` function. It requires two main arguments:
1. **File Name / Path:** The name or location of the file.
2. **Mode:** A character that tells Python what you want to do with the file.

| Mode | Description |
| :---: | :--- |
| **`'r'`** | **Read (Default):** Opens a file for reading. Throws an error if the file does not exist. |
| **`'w'`** | **Write:** Opens a file for writing. Creates a new file if it doesn't exist, or **overwrites/erases** all existing content. |
| **`'a'`** | **Append:** Opens a file to add data at the end. Creates a new file if it doesn't exist. It **does not** delete old data. |
| **`'x'`** | **Create:** Creates a new file. Throws an error if the file already exists. |

# **Reading to a Text File "r"**

````python
file_1 = open("sample_1.txt", "r")
# print(type(file_1))
print(file_1.read())
file_1.close()

file_2 = open("sample_2.txt", "r")
print(file_2.read())
````

## **The Safe Way to Read a File**

- This breakdown explains how Python handles file reading securely and automatically using a **Context Manager** (`with` statement).## Code Breakdown: The Safe Way to Read a File

- This breakdown explains how Python handles file reading securely and automatically using a **Context Manager** (`with` statement).

````python
with open("sample_1.txt", "r") as file_1:
    print(file_1.read())
````

## **Reading the First Line of a File**
This breakdown explains how Python safely opens a file and extracts only its **very first line** using the `readline()` method.

````python
with open("sample_1.txt", "r") as file_1:
    line_1 = file_1.readline()
    print(line_1)
    
````

## **Reading All Lines using a loop**

````python
with open("sample_1.txt", "r") as file_1:
    while True:
        line_1 = file_1.readline()
        if not line_1:
            break
        print(line_1)
    
````

## **Reading All Lines into a List**

This breakdown explains how Python safely opens a file and extracts all of its lines simultaneously into a structured **Python List** using the `readlines()` method.

````python
with open("sample_1.txt","r") as file_1:
    content = file_1.readlines()
    print(content)
    print(type(content))
````

# **Writing to a Text File "w"**

- This breakdown explains how Python handles creating a fresh file and writing multiple lines of text into it using the **Write mode (`"w"`)**.

- **What Happens When You Run These Two Cells Sequentially?**

    - If you execute the first code cell and then execute the second code cell, your final file (`new_sample.txt`) will **ONLY contain the text from the second cell**. 

    - The text from the first cell will be **completely deleted (overwritten)**.

````python
with open("new_sample_1.txt","w") as file_1:
    file_1.write("This is a new line 1")
    file_1.write("\nThis is a new line 2")
````

````python
with open("new_sample_1.txt","w") as file_1:
    file_1.write("This is a new line 11")
    file_1.write("\nThis is a new line 22")
````

### **Writing a List of Lines using `writelines()`**

The `writelines()` method is used when you have a collection of data (like a **List of strings**) and you want to write all those items into a file at once.

| Feature | `write()` 📝 | `writelines()` 📚 |
| :--- | :--- | :--- |
| **Primary Purpose** | Writes a **single** string to the file. | Writes a **list (collection)** of strings to the file at once. |
| **Accepted Input Type** | `str` (String only) | An iterable container of strings (`list`, `tuple`, `set`, etc.) |
| **Automatic Newlines (`\n`)?**| **No.** You must manually add `\n` to break lines. | **No.** You must manually add `\n` inside the list items. |
| **Raises TypeError if...** | You pass a list or any non-string type. | You pass a single integer or a list containing non-string items. |

````python
with open("new_sample_2.txt","w") as file_2:
    file_2.writelines([
        "This is a new line 1",
        "\nThis is a new line 2",
        "\nThis is a new line 3",
        "\nThis is a new line 4"
        ])
````

---

# **Append "a"**

- The Append mode (**`"a"`**) is used when you want to add new data to the end of an existing file **without losing or erasing** any of its current content.

````python
with open("new_sample_2.txt","a") as file_2:
    file_2.writelines([
        "\nThis is a new line 5"
        "\nThis is a new line 6"
    ])
````

# **Exclusive Creation "x"**

- The Exclusive Creation mode (**`"x"`**) is a highly secure mode used when you want to create and write to a **brand new file**, but you want to ensure you **never accidentally overwrite** an existing file with the same name.

````python
with open("new_sample_1.txt","x") as file_1:
    file_1.write("This is a new line 1")
    file_1.write("\nThis is a new line 2")
    file_1.write("\nThis is a new line 3")
    file_1.write("\nThis is a new line 4")
    file_1.write("\nThis is a new line 5")
    file_1.write("\nThis is a new line 6")
    file_1.write("\nThis is a new line 7")
    file_1.write("\nThis is a new line 8")
    file_1.write("\nThis is a new line 9")
    file_1.write("\nThis is a new line 10")
````
