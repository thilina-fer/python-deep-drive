# **JSON  (Java Script Onbject Notation)**

JSON  data is a represented as **a collection of key-value pairs**, where **keys are strings** enclosed in double quation marks and values can be strings , numbers , booleans , arrays or the JSON objects

| Feature | Python Dictionary (`dict`) 🐍 | JSON (JavaScript Object Notation) 🌐 |
| :--- | :--- | :--- |
| **What is it?** | A native, executable **Data Structure** in Python memory. | A text-based **Data Interchange Format** (String text). |
| **Syntax Scope** | Specific to Python programming language. | Language-independent (Supported by almost all languages). |
| **Keys Type** | Can be strings, integers, tuples, or any hashable object. | **Must be strings only**, enclosed in double quotes (`""`). |
| **String Quotes** | Can use single quotes (`'`) or double quotes (`"`). | **Must use double quotes (`"`)** for keys and strings. |
| **Booleans** | Written as `True` or `False`. | Written as lowercase `true` or `false`. |
| **Empty/Null Value** | Uses `None`. | Uses `null`. |
| **Trailing Commas**| Allowed at the end of the last item. | ❌ **Strictly forbidden**; will throw a parsing error. |


* The line **import json** in python used to import json linrary, which provides functionality for working with JSON data 

---


- Python reads a JSON file and automatically converts its text-based format into a native Python data structure using the `json.load()` method.

```python
import json

with open("sample_1.json", "r") as file :
    data = json.load(file)
    print(data)
    print(type(data))
```

```python
import json 

with open("sample_2.json","r") as file :
    data_2 = json.load(file)
    print(data_2)
    print(type(data_2))
```

# **json.dump() method**

- dump is used to save a python object directly into a JSON file.It converts the data to JSON format and writes it to the file all in one step 

```python
import json

data = [{'id': 1, 'name': 'John', 'age': 30}, {'id': 2, 'name': 'Jane', 'age': 25}]

with open("sample_3.json", "w") as file:
    json.dump(data, file, indent=4)
```
