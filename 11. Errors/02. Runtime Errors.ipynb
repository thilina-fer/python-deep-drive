# **Runtime Errors**

A runtime error in Python is an exception that occurs while your program is executing, after it has successfully passed the syntax checks. Unlike syntax errors which prevent a program from starting, runtime errors cause a running application to crash abruptly unless they are properly intercepted and managed.

Common Types of Runtime Errors

- Python classifies different runtime issues into specific built-in exceptions. Here are the most frequent culprits:
    - **ZeroDivisionError**: Trying to divide a number by zero.
    - **NameError**: Referencing a variable or a function name that has not been defined yet.
    - **TypeError**: Combining incompatible data types, like adding a string to an integer.
    - **ValueError**: Passing an argument with the correct type but an inappropriate value, such as converting "apple" into an integer.
    - **IndexError**: Trying to access an out-of-bounds index in a list or a tuple.
    - **KeyError**: Attempting to look up a dictionary key that does not exist.FileNotFoundError: Opening or reading a file path that is missing.

### **How to Handle Runtime Errors**
- Instead of allowing your application to crash, you can safely intercept exceptions using try and except blocks. This technique allows you to log the issue, alert the user, or run fallback code safely

```python
try:
    # Code that might cause a runtime error
    number = int(input("Enter a number: "))
    result = 100 / number
except ValueError:
    # Catches non-numeric inputs
    print("Please enter a valid integer.")
except ZeroDivisionError:
    # Catches an input of 0
    print("Division by zero is not allowed.")
else:
    # Runs only if no exceptions occur
    print(f"Success! The result is {result}")
finally:
    # Always runs, useful for clean-up tasks
    print("Execution complete.")
```
