# **Python Debugging: Core Principles & Essential Techniques**

## **Core Debugging Principles**
* **Reproduce the error reliably:** Find the exact steps or inputs that trigger the bug consistently.
* **Isolate the problem:** Narrow down the code to the smallest possible snippet that still fails.
* **Check assumptions:** Verify that your variables, inputs, and functions actually contain what you think they contain.
* **Fix the cause, not the symptom:** Avoid "band-aid" fixes that patch a specific crash but leave the underlying logic broken.

## **Essential Techniques & Tools**

### **1. Strategic Print Statements**
* Use `print()` or f-strings to inspect variable states.
* Leverage the `repr()` function or `f"{variable=}"` syntax to see exact data types and hidden characters (like whitespaces).

### **2. Specialized Logging**
* Replace `print()` with Python’s built-in `logging` module.
* Set different severity levels (`DEBUG`, `INFO`, `WARNING`, `ERROR`, `CRITICAL`).
* Disable or redirect outputs easily without changing your core logic.

### **3. Interactive Debugging (PDB)**
* Insert `breakpoint()` directly into your code to pause execution.
* Use commands like `n` (next line), `s` (step into function), `c` (continue execution), and `p` (print variable) to inspect the live environment.

### **4. IDE Debuggers**
* Use visual debuggers in PyCharm or VS Code.
* Set conditional breakpoints that only trigger when a variable hits a specific value.
* Watch variables change in real-time using a graphical user interface.

## **Common Python Pitfalls to Check**
* **Mutable Default Arguments:** Avoid using lists or dictionaries as default arguments in functions (e.g., `def func(x=[]):`).
* **Scope Confusion:** Ensure you are not accidentally shadowing global variables inside a local function block.
* **Type Mismatches:** Python is dynamically typed; verify you are not trying to concatenate a string with an integer.
