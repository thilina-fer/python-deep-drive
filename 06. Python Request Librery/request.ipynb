**Basic HTTP Methods**
The library maps standard HTTP verbs directly onto intuitive Python functions:
- **GET**: Retrieves data from a specified URL.
- **POST**: Submits or creates new data payload to a server.
- **PUT**: Updates an entire existing web resource.
- **DELETE**: Deletes a specific resource on the target server.

````python
import requests

response = requests.get('https://api.github.com/events')
print(response.status_code)
print(response.headers)
print(response.headers['Content-Type'])
````

````python
if "application/json" in response.headers['Content-Type']:
    print("It is a JSON response")
else :
    print("It is not a JSON response")
````

## **response.text**

- In Python's requests library, response.text is a property that returns the HTTP response body decoded as a Unicode string.
- When you access response.text, the requests library automatically guesses the character encoding based on the HTTP headers sent by the server and decodes the underlying raw payload for you.

**Key Behaviors**
- **It is a property, not a method**: Do not use parentheses like response.text(). Doing so will throw a TypeError: 'str' object is not callable.
- **Automatic Decoding**: It inspects the Content-Type header's charset to handle the decoding. If no encoding is specified by the server, it defaults to ISO-8859-1 per RFC 2616 specifications, or uses fallback libraries like charset-normalizer to guess the layout.
- **Overriding Encoding**: If you notice that your characters are garbled or decoded improperly, you can force a specific encoding before calling 

````python
print(response.text)
print(type(response.text))
````

## **response.content**

- In the Python requests library, response.content returns the raw, unencoded byte string (the bytes data type) of the server's response body. It is primarily used when downloading non-text assets like images, PDFs, zip files, or when you need to manually handle character encoding decoding.

````python
print(response.content)
print(type(response.content))
````

### **Key Differences: `content` vs. `text` vs. `json()`**

| Attribute / Method | Data Type Returned | Best Used For |
| :--- | :--- | :--- |
| `response.content` | `bytes` | Binary files (images, video streams, audio, PDFs, executables). |
| `response.text` | `str` (Unicode) | Text-based web pages (HTML, XML, plain text) where `requests` automatically guesses the character encoding. |
| `response.json()` | `dict` or `list` | REST APIs that return data formatted as a JSON payload. |

````python
# Converts JSON to Dict
if "application/json" in response.headers['Content-Type']:
    data_dict = response.json()
    print(data_dict)
    print(type(data_dict))
else :
    print("It is not a JSON response")
````
