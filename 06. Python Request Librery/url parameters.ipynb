## **query parameters**

In the Python requests library, you pass query parameters to a URL by using the params argument inside request methods like requests.get(). The library automatically handles URL encoding, formatting (adding ? and &), and mapping your key-value pairs.

**Advanced Parameter Types**
- Beyond simple dictionaries, the requests library provides flexible handling for more complex query string requirements:
    - Sending Multiple Values for One Key: If an API requires multiple values for the same parameter (e.g., ?tags=python&tags=api), pass those values inside a list.
    - Handling Duplicate Keys with Lists of Tuples: If order matters or you want to explicitly avoid dictionary formatting, pass a list of two-item tuples.
    - Ignoring None Values: Any key assigned a value of None will be completely omitted from the final URL query string.

```python
import requests
url = "https://jsonplaceholder.typicode.com/posts"

params = {'userId':1, 'id':2}
response = requests.get(url, params=params)
print(response.url)
print(response.text)
```
