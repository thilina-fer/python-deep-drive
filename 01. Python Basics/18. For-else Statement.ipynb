# The `for-else` Statement

The `else` block in a `for` loop has a specific logic: It executes **only if the loop finishes normally**. If the loop is terminated by a `break` statement, the `else` block is skipped.

**Common Use Case:**
It is most useful for "search" operations. You use the `for` loop to look for something; if you find it, you `break`. If the loop finishes and you never hit the `break`, the `else` block runs to handle the "not found" scenario.

```python
numbers = [1, 2, 3, 4, 5]
search_for = 3

for n in numbers:
    if n == search_for:
        print("Found the number!")
        break # This prevents the 'else' block from running
else:
    # This runs only if the loop finishes without finding the number
    print("Search complete: The number was not in the list.")
```
