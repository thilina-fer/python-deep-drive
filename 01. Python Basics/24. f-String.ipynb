# **f-String**

- f-Strings (Formatted String Literals), introduced in Python 3.6, are arguably the most popular and efficient way to format strings. They are faster, more readable, and more concise than the older .format() or % methods.
- To create an f-string, you simply prefix the string with an f or F and use curly braces {} to evaluate expressions directly within the text.

```python
name = "Alice"
age = 25

# Simple interpolation
print(f"Hello, {name}. You are {age} years old.")
# Output: Hello, Alice. You are 25 years old.
```
