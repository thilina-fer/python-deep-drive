# **Lambda Functions (Anonymous Functiosnn)**

- A Lambda Function is a small, anonymous function in Python. It is called "anonymous" because, unlike a standard function defined with def, a lambda function doesn't need a name.

- It is designed for one-time use or for cases where you need a simple piece of logic for a short period.

- `lambda arguments: expression`
    - lambda: The keyword that starts the function.
    - arguments: The inputs (can be one or many, separated by commas).
    - expression: A single piece of code that is executed and automatically returned.



```python
addition = lambda a,b : a+b
print(addition(3,4))
```

```python
maximum_number = lambda a,b : a if a > b else b
print(maximum_number(10,28))
```

If you are using a lamda and you give it a name , you are doing it wrong. Use def. Only use a lambda if you need to pass a tiny, one-line function into another function(like sorted) and you are never going to use it again

### When to Use `def` vs. `lambda`

| Situation | Use `def` | Use `lambda` |
| :--- | :--- | :--- |
| **Logic is complex** | ✅ Always | ❌ No |
| **Used multiple times** | ✅ Always | ❌ No |
| **Assigned to a name** | ✅ Always | ❌ No (Anti-pattern) |
| **One-line, throwaway** | 🆗 Optional | ✅ Best |
| **As an argument for `sorted()`** | 🆗 Optional | ✅ Best |

- **If you have to give it a name to use it later, it’s not a lambda; it’s a function. Use def.**

```python
users = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 25},
    {"name": "Charlie", "age": 35},
    ]

# Sort users by age
sorted_users = sorted(users, key=lambda user: user["age"])
print(sorted_users)
```

```python
# Bad Practice(Violates PEP 8 style guide)
calculate_tax = lambda price : price * 0.07
print(calculate_tax(100))

# Good Practice
def calculate_tax(price):
    return price * 0.07
print(calculate_tax(100))
```

```python
users = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 25},
    {"name": "Charlie", "age": 35},
    ]
maximum_age = max(users, key= lambda user: user["age"])
print(maximum_age)

minimum_age = min(users, key= lambda user: user["age"])
print(minimum_age)
```

```python
# Not Recomended(Violates PEP 8 style guide)

my_list = [1,2,3,4,5]
result = list(map(lambda x: x**2, my_list))
print(result)
```
