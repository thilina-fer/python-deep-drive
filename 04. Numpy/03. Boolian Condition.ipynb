# **Boolean Conditions** 

- Element-wise evaluation: NumPy checks each number individually (\(10>15\), \(20>15\), etc.).
- Data Type: The resulting condition variable is a standard NumPy array (numpy.ndarray).
- Internal Data Type: The elements inside this new array are booleans (bool_).

```python
# Boolean Comparison in Numpy Arrays
import numpy as np

arr = np.array([10,20,30,14,15])
condition = arr > 15
print(condition)
print(type(condition))
```

```python
# Boolean Indexing in Numpy Arrays
import numpy as np

arr = np.array([10,20,30,14,15])
condition = arr > 15

filtered_array = arr[condition]
print(filtered_array)
```

```python
import numpy as np

arr = np.array([10,20,30,14,15])

filtered_array = arr[arr > 15]
print(filtered_array)

filterd_array_2 = arr[(arr > 10) & (arr < 25)]
print(filterd_array_2)
```
