## List Comprehension

List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list. It is faster and more readable than using a traditional `for` loop with `.append()`.

**Basic Syntax:**
`new_list = [expression for item in iterable if condition == True]`

* **Expression**: The operation or value to be added to the new list.
* **Item**: The temporary variable representing each element in the sequence.
* **Iterable**: The source sequence (list, range, etc.).
* **Condition** (Optional): A filter to decide if the item should be included.

````python
#The Traditional Way
numbers = [1, 2, 3, 4, 5]
squares = []

for x in numbers:
    squares.append(x**2)

print(squares) # Output: [1, 4, 9, 16, 25]
````

````python
# The List Comprehension Way
numbers = [1, 2, 3, 4, 5]
squares = [x**2 for x in numbers]

print(squares) # Output: [1, 4, 9, 16, 25]


my_list = [1, 2, 3, -4, -5]
positive_square = [i**2 for i in my_list if i > 0]
print(positive_square) # Output: [1, 4, 9]
````

Write a python program to calculate the total value of current inventory and store it in the total_store_value variable .

# The `break` Statement

The `break` statement is used to exit a `for` loop or a `while` loop before it has finished iterating through the entire sequence.

**Key Characteristics:**
* **Immediate Termination**: It stops the loop as soon as the condition is met.
* **Control Flow**: It "breaks" out of the current loop and jumps to the code following the loop block.
* **Common Use**: Often used inside an `if` statement to stop a search once a specific item is found.

````python
for i in range(1, 11):
    if i==6:
        break
    print(i)
````

# The `continue` Statement

The `continue` statement is used to skip the current iteration of a loop and move immediately to the next one.

**How it works:**
* When Python hits `continue`, it ignores any code written below it inside that loop block.
* It jumps back to the top of the loop to start the next turn.
* **Use Case**: This is perfect when you want to "filter out" or ignore specific values without stopping the whole process.
