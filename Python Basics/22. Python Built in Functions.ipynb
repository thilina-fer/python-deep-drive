## **1. abs()**
 - The abs() function is a built-in Python function used to calculate the absolute value of a number. In mathematical terms, it returns the distance of a number from zero on a number line, regardless of its direction (positive or negative).


| Input Type | Example | Result | Description |
| :--- | :--- | :--- | :--- |
| **Integer** | `abs(-10)` | `10` | Removes the negative sign. |
| **Float** | `abs(-33.5)` | `33.5` | Removes the negative sign. |
| **Positive** | `abs(5)` | `5` | No change. |
| **Complex** | `abs(3 + 4j)` | `5.0` | Returns the **magnitude** of the number. |

````python
print(abs(-5))      # Output: 5
print(abs(5))       # Output: 5
print(abs(-1.23))   # Output: 1.23
print(abs(3+4j))    # Output: 5.0
````

## **2. Map()**
- The map() function is a built-in Python tool used to apply a specific function to every item in an iterable (like a list or tuple) without using a manual for loop.

``` map(function, iterable) ```
- function: The logic you want to apply to every item.
- iterable: The collection of data (list, tuple, etc.) you want to transform.

- When you call map(), Python doesn't immediately create a new list. It returns an iterator. This is memory efficient because it only calculates the transformation when you actually ask for the item (Lazy Evaluation). To see the results clearly, you usually wrap it in list().

#### Comparison: `map()` Function vs. Traditional `for` Loop

| Feature | `map()` Function | Traditional `for` Loop |
| :--- | :--- | :--- |
| **Code Length** | Short and concise (one line) | Longer (multiple lines) |
| **Performance** | Faster (implemented in C) | Slightly slower in Python |
| **Readability** | High for simple transformations | High for complex logic |
| **Memory** | Efficient (uses an iterator) | Usually creates the whole list in memory |



````python
def find_square(x):
    return x**2
my_list = [1, 2, 3, 4, 5]
result = map(find_square, my_list)
print(result)  # Output: <map object at 0x7f8b8c8d9e80>
print(type(result))
print(list(result))  # Output: [1, 4, 9, 16, 25]
````

#### 1. Converting Data Types `(int , str , float)`
- This is most common and argubably the best use for `map()`.It clearly converts a lis of one data type ito another .

````python
user_inputs = ["10", "20", "30"]
converted_inputs = map(int, user_inputs)
print(list(converted_inputs))  # Output: [10, 20, 30]
print(type(converted_inputs))  # Output: <class 'map'>
````

#### 2. Getting the length of items `(len())`

If you hava a list of strings(or a list of lists) and you want to knaw the size of each one, `map()` handles this insrantly.

````python
words = ["hello", "world", "python"]
word_length = map(len, words)
print(list(word_length))  # Output: [5, 5, 6]
````

#### 3. Quick math operations `(abs,round)`

- You can apply standard built-in math functions across an entire dataset without writting any loops

````python
transactions = [-50.2 , 56.3 ,100.1 , -20.5]
positive_numbers = list(map(abs, transactions))
print(positive_numbers)  # Output: [50.2, 56.3, 100.1, 20.5]

floating_numbers = [1.5, 2.3, 3.7]
rounded_numbers = list(map(round, floating_numbers))
print(rounded_numbers)  # Output: [2, 2, 4]
````

#### 4. Working with multiiple `(max,min,pow)`

A lesser known superpower of map() is that it can take multiple list at the same time , as long as built in functions accept multiple arguments.


````python
list_1 = [75,32,94,55,85]
list_2 = [60,53,22,36,94]

max_values = list(map(max, list_1, list_2))
min_values = list(map(min, list_1, list_2))
print(max_values)
print(min_values)
````

## **3. Sum()**

The sum() function is a built-in Python tool used to add up all the items in an iterable (like a list, tuple, or set).
- `sum(iterable, start)`
    - **iterable**: The collection of numbers you want to add.
    - **start (Optional**): A value that is added to the total. It defaults to 0.

````python
numbers = [1, 2, 3, 4]
print(sum(numbers)) # Output: 10
````

## **4. isinstance()**

The isinstance() function is used to check if an object belongs to a specific class or a tuple of classes. It is the preferred way to perform type checking in Python because it supports inheritance (it returns True if an object is an instance of a subclass).

- `isinstance(object, classinfo)`

    - **object** : The instance you want to check.
    - **classinfo** : A class, type, or a tuple containing multiple classes/types.

````python
x = 10
print(isinstance(x, int))   # True
print(isinstance(x, str))   # False

y = 3.14
print(isinstance(y, (int, float))) # True (It's a float)
````

### `isinstance()` vs. `type()` Comparison

| Feature | `isinstance()` | `type()` |
| :--- | :--- | :--- |
| **Purpose** | Checks if object is an instance of a class. | Returns the exact type of an object. |
| **Inheritance** | **Supports it** (Returns True for parent classes). | **Strict** (Only True for the exact class). |
| **Multi-check** | Supports a tuple of classes. | Usually requires multiple `==` checks. |
| **Best Practice** | Recommended for most logic checks. | Best when you need the *exact* class only. |

````python
class Animal:
    pass

class Dog(Animal):
    pass

my_dog = Dog()

# isinstance() checks subclasses too
print(isinstance(my_dog, Animal)) # True

# type() checks the exact class only
print(type(my_dog) == Animal)     # False
````
