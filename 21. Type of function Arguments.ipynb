# Type of Python function Arguments

In Python, the way you pass Arguments to a function determines how the Parameters receive those values. Understanding these four types is essential for writing flexible code.

1. Positional Arguments
2. Default Arguments
3. Keyword Arguments
4. Arbitrary Arguments (*args and kwargs)

## 1. Positional Arguments
- These are the most basic types of arguments. The values are assigned to parameters based on their order (position).


```python
def my_function(a,b):
    print(a)
    print(b)
my_function(10,20)
```

## 2. Default Arguments
- You can provide a default value to a parameter in the function definition. If no argument is provided during the function call, the default value is used.

```python
def my_function(a,b=5):
    print(a,b)
    
my_function(10)
my_function(10,20)
my_function(10,20,30)
```
