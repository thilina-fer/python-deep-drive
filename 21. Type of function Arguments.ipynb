# Type of Python function Arguments

In Python, the way you pass Arguments to a function determines how the Parameters receive those values. Understanding these four types is essential for writing flexible code.

1. Positional Arguments
2. Default Arguments
3. Keyword Arguments
4. Arbitrary Arguments (*args and kwargs)

## 1. Positional Arguments
- These are the most basic types of arguments. The values are assigned to parameters based on their order (position).


````python
def my_function(a,b):
    print(a)
    print(b)
my_function(10,20)
````

## 2. Default Arguments
- You can provide a default value to a parameter in the function definition. If no argument is provided during the function call, the default value is used.

````python
def my_function(a,b=5):
    print(a,b)
    
my_function(10)
my_function(10,20)
my_function(10,20,30)
````

**Q.**
Write a function call power that accept 2 paremeters, base and exponent, and returns the base raise to the power of the exponent .If the exponent the not provided it should default to 2 

````python
def power(base, exponent=2):
    return base ** exponent

print(power(2, 3))  
print(power(5)) 
print(power(0))
````

**The "Non-Default Last" Rule**
- In Python, once you provide a default value to a parameter (e.g., b=5), every parameter that follows it in the function definition must also have a default value. You cannot have a "required" (non-default) parameter trailing behind an "optional" (default) one.

````python
def function(a, b=5, c=10):
    print(a, b, c)

function(1)
````

````python
def function(a=10 , b):
    print(a, b)
function(1)
````

## 3. Keyword Argument
- Keyword Arguments (often abbreviated as kwargs) allow you to pass arguments to a function by explicitly naming the parameter. This makes your code more readable and allows you to ignore the order of arguments.

````python
def my_function(a,b,c):
    print("a:",a)
    print("b:",b)
    print("c:",c)
# my_function(a=1, c=3, b=2)
# my_function(c=3, a=1, b=2)
my_function(b=2, c=3, a=1)
````

## 4. Arbitrary Argiments

Mainly there are 2 typpes 

1. Arbitrary Positional Arguments -> (*args)
2. Arbitrary Keyword Arguments -> (**kwargs)

### 4.1. Arbitrary Positional Arguments (*args)

- If we dont know the num of parameter conunt then we can use Arbitrary Positional Arguments 
- Arbitrary Positional Arguments allow a function to accept any number of inputs. This is extremely useful when you don't know beforehand how many values a user might pass into your function.
- By adding an asterisk (*) before a parameter name, you tell Python to collect all remaining positional arguments into a Tuple.

- def func(a, b, *args):  ->  Valid
- def func(*args, a, b):  ->  Invalid

````python
def my_function(*args):
    print(args)
    print(type(args))
my_function(1, 2, 3, 4, 5)
````

````python
def my_function(**args):
    print(args)
    print(type(args))
my_function(a=1, b=2, c=3)
````

**Q.** Write a python function transform_sequence that accept
- An arbitrary number of integers
- A keyword only parameter mode with default value  "liner"
- **Rule** : Your function should transform each number based on the selected mode
1. Linear mode f(x) = 2x + 3
2. Quadratic mode f(x) = x^2 - x
3. Factorial_sum mode for each number x compute sum of factorial 1 to x  ->  f(r) = 1! + 2! + 3! + 4! + ... + x!  .(Assium x >= 1)
Function output should return a dictionary as below :
```
{
    "transformed : [...],
    "total" : ...,
    "max" : ...,
    "min" : ...
} 
``` 
```
Teat cases : 
transforme_sequence(1,12,3,4)
transforme_sequence(1,2,3,mode = "quadratic")
transforme_sequence(3,4,5, mode = "factorial_sum")
```

````python
import math

def transform_sequence(*value , mode = "liner"):
    transformed = []
    match mode :
        case "liner":
          transformed = [v*2+3 for v in value]
        case "quadratic":
          transformed = [v**2-v for v in value]
        case "sum_of_factorial":
          transformed = [
                sum(math.factorial(i) for i in range(1, v + 1))
                for v in value
            ]

    return {
       "transformed": transformed,
       "total": sum(transformed),
       "max": max(transformed),
       "min": min(transformed)
    }

print(transform_sequence(1,2,3))
print(transform_sequence(1,2,3, mode="liner"))
print(transform_sequence(1,2,3, mode="quadratic"))
print(transform_sequence(1,2,5, mode="sum_of_factorial"))
        
````

### 4.2. Arbitrary Keyword Arguments(**keyword)
 - Arbitrary Keyword Arguments are special type for prameter that **allows a function to accept any number of keyword arguments.** The double asterisk(**) is used to donate an arbitrary keyword argument.


````python
def my_function(**kwargs):
    print(kwargs)
    print(type(kwargs))
    for key, value in kwargs.items():
        print(f"{key}: {value}")
my_function(name = "Alice", age = 30, city = "New York")
````
