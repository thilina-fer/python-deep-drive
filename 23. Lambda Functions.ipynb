# **Lambda Functions (Anonymous Functiosnn)**

- A Lambda Function is a small, anonymous function in Python. It is called "anonymous" because, unlike a standard function defined with def, a lambda function doesn't need a name.

- It is designed for one-time use or for cases where you need a simple piece of logic for a short period.

- `lambda arguments: expression`
    - lambda: The keyword that starts the function.
    - arguments: The inputs (can be one or many, separated by commas).
    - expression: A single piece of code that is executed and automatically returned.



```python
addition = lambda a,b : a+b
print(addition(3,4))
```

```python
maximum_number = lambda a,b : a if a > b else b
print(maximum_number(10,28))
```
