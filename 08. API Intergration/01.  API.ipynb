```python
import requests

url = "https://api.open-meteo.com/v1/forecast"

params={
    "latitude" : "6.7118",
    "longitude" : "79.9051",
    "current" : "temperature_2m,wind_speed_10m"
}

response = requests.get(url, params=params)

match response.status_code:
    case 200:
        data = response.json()
        current_data = data["current"]
        print(data)
        temparature = current_data["temperature_2m"]
        wind  = current_data["wind_speed_10m"]
        print(f"Temparature : {temparature}")
        print(f"Wind Speed : {wind}")
    case _ :
        print("Error")
```
